# Comparing Distributions Across Multiple Columns
## Data Science Milestone: Exploratory Data Analysis (EDA)

**Objective:** Learn to compare distributions across multiple columns in a Pandas DataFrame to understand how different variables behave relative to each other.

**Key Skills:**
- Compute summary statistics for multiple columns
- Compare central tendency (mean, median)
- Compare spread and variability (std, range, IQR)
- Identify patterns and anomalies
- Interpret comparative statistics meaningfully

**Why This Matters:**
- Most real insights come from comparison, not isolation
- Distribution comparison reveals hidden patterns
- Informs better analysis decisions
- Prevents misleading conclusions from single-column analysis

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Suppress warnings for cleaner output
import warnings
warnings.filterwarnings('ignore')
```

## Section 1: Load and Explore the DataFrame

First, we load the data and understand its structure, data types, and shape.

```python
# Load the CSV file
csv_path = Path("../data/raw/sample_people.csv")
df = pd.read_csv(csv_path)

# Display basic information
print("=" * 80)
print("DATAFRAME STRUCTURE")
print("=" * 80)
print(f"\nShape: {df.shape[0]} rows × {df.shape[1]} columns")
print("\nFirst few rows:")
print(df)
print("\nData types:")
print(df.dtypes)
print("\nBasic info:")
print(df.info())
```

## Section 2: Compute Summary Statistics for Multiple Columns

Using `describe()`, we get comprehensive statistics for all numeric columns at once.

```python
# Get summary statistics for ALL numeric columns
print("=" * 80)
print("SUMMARY STATISTICS FOR ALL NUMERIC COLUMNS")
print("=" * 80)
summary_stats = df.describe()
print("\n", summary_stats.round(2))

# What each statistic means:
print("\n\n[UNDERSTANDING EACH STATISTIC]")
print("- count: Number of non-null values")
print("- mean: Average value (sum ÷ count)")
print("- std: Standard deviation (measure of spread)")
print("- min: Smallest value")
print("- 25%: First quartile (Q1) - 25% of data is below this")
print("- 50%: Median (Q2) - middle value")
print("- 75%: Third quartile (Q3) - 75% of data is below this")
print("- max: Largest value")
```

## Section 3: Compare Central Tendency Across Columns

Central tendency tells us where the "center" of each distribution is. We compare means and medians.

```python
# Extract and compare means and medians
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

print("=" * 80)
print("COMPARING CENTRAL TENDENCY: MEANS AND MEDIANS")
print("=" * 80)

central_comparison = pd.DataFrame({
    'Column': numeric_cols,
    'Mean': [df[col].mean() for col in numeric_cols],
    'Median': [df[col].median() for col in numeric_cols],
    'Mean - Median': [df[col].mean() - df[col].median() for col in numeric_cols]
})

print("\n", central_comparison.round(2))

# Interpretation
print("\n\n[INTERPRETING THE DIFFERENCES]")
for col in numeric_cols:
    mean_val = df[col].mean()
    median_val = df[col].median()
    diff = mean_val - median_val
    
    print(f"\n'{col}' Column:")
    print(f"  Mean: {mean_val:.2f}")
    print(f"  Median: {median_val:.2f}")
    
    if abs(diff) < 1:
        print(f"  → Mean ≈ Median: Distribution is fairly SYMMETRIC")
    elif diff > 0:
        print(f"  → Mean > Median: Distribution is RIGHT-SKEWED")
        print(f"     (Some unusually high values pull the average up)")
    else:
        print(f"  → Mean < Median: Distribution is LEFT-SKEWED")
        print(f"     (Some unusually low values pull the average down)")
```

## Section 4: Compare Spread and Variability Across Columns

Spread and variability tell us how scattered or concentrated the data is. We compare ranges, standard deviations, and quartile ranges.

```python
# Compare spread and variability
print("=" * 80)
print("COMPARING SPREAD AND VARIABILITY")
print("=" * 80)

variability_comparison = pd.DataFrame({
    'Column': numeric_cols,
    'Min': [df[col].min() for col in numeric_cols],
    'Max': [df[col].max() for col in numeric_cols],
    'Range': [df[col].max() - df[col].min() for col in numeric_cols],
    'Q1 (25%)': [df[col].quantile(0.25) for col in numeric_cols],
    'Q3 (75%)': [df[col].quantile(0.75) for col in numeric_cols],
    'IQR': [df[col].quantile(0.75) - df[col].quantile(0.25) for col in numeric_cols],
    'Std Dev': [df[col].std() for col in numeric_cols],
})

print("\n", variability_comparison.round(2))

# Identify high vs low variability
print("\n\n[INTERPRETING VARIABILITY]")
std_values = {col: df[col].std() for col in numeric_cols}
max_std_col = max(std_values, key=std_values.get)
min_std_col = min(std_values, key=std_values.get)

print(f"\n✓ HIGHEST VARIABILITY: '{max_std_col}' (Std Dev: {std_values[max_std_col]:.2f})")
print(f"  → Data is more SCATTERED and less consistent")
print(f"  → Values vary widely from the mean")

print(f"\n✓ LOWEST VARIABILITY: '{min_std_col}' (Std Dev: {std_values[min_std_col]:.2f})")
print(f"  → Data is more CONCENTRATED and more stable")
print(f"  → Values stay close to the mean")

# Coefficient of Variation (relative variability)
print("\n\n[COEFFICIENT OF VARIATION (Relative Variability)]")
print("CV% = (Standard Deviation / Mean) × 100")
print("This shows variability relative to the mean on a fair scale.\n")

for col in numeric_cols:
    cv = (df[col].std() / df[col].mean()) * 100
    print(f"{col:10} CV: {cv:6.1f}%", end="")
    if cv < 15:
        print(" → Low relative variation (stable variable)")
    elif cv < 30:
        print(" → Moderate relative variation")
    else:
        print(" → High relative variation (volatile variable)")
```

## Section 5: Identify Patterns and Anomalies

We use statistics to detect unusual behavior, outliers, and patterns that warrant deeper investigation.

```python
# Identify patterns and anomalies using quartile-based outlier detection
print("=" * 80)
print("IDENTIFYING PATTERNS AND ANOMALIES")
print("=" * 80)

for col in numeric_cols:
    print(f"\n\n[{col.upper()} COLUMN ANALYSIS]")
    print("-" * 80)
    
    # Quartile analysis
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    
    print(f"Q1 (25th percentile): {q1:.2f}")
    print(f"Q3 (75th percentile): {q3:.2f}")
    print(f"IQR (middle 50% range): {iqr:.2f}")
    
    # Outlier detection (1.5 × IQR rule)
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    print(f"\nOutlier bounds: [{lower_bound:.2f}, {upper_bound:.2f}]")
    
    if len(outliers) > 0:
        print(f"⚠ Potential outliers detected: {len(outliers)} values")
        for idx, row in outliers.iterrows():
            print(f"   - {row['Name']}: {col} = {row[col]}")
    else:
        print(f"✓ No outliers detected")
    
    # Skewness
    skewness = df[col].skew()
    print(f"\nSkewness: {skewness:.3f}", end="")
    if abs(skewness) < 0.5:
        print(" → Fairly symmetric distribution")
    elif skewness > 0.5:
        print(" → Right-skewed distribution (tail on right, higher values pull mean up)")
    else:
        print(" → Left-skewed distribution (tail on left, lower values pull mean down)")
```

## Section 6: Interpret and Document Findings

Let's synthesize all our learning and document key insights.

```python
# Correlation analysis - comparing relationships
print("\n\n" + "=" * 80)
print("CORRELATION ANALYSIS: HOW VARIABLES RELATE")
print("=" * 80)

corr_matrix = df[numeric_cols].corr()
print("\nCorrelation between all numeric columns:")
print(corr_matrix.round(3))

# Get Age-Salary correlation specifically
if 'Age' in numeric_cols and 'Salary' in numeric_cols:
    correlation = df['Age'].corr(df['Salary'])
    print(f"\n✓ Age-Salary Correlation: {correlation:.3f}")
    if correlation > 0.7:
        print("  → Strong positive relationship: Older people tend to have higher salaries")
    elif correlation > 0.3:
        print("  → Moderate positive relationship: Some tendency for salaries to increase with age")
    else:
        print("  → Weak relationship: Age and salary don't move together strongly")

# Prepare for deeper insights
print("\n\n" + "=" * 80)
print("Z-SCORE NORMALIZED COMPARISON")
print("=" * 80)
print("(Standardized scores put both columns on the same scale)")

df_zscore = df[numeric_cols].copy()
for col in numeric_cols:
    df_zscore[col] = (df[col] - df[col].mean()) / df[col].std()

comparison_df = pd.DataFrame({
    'Name': df['Name'],
    'Age': df['Age'],
    'Age_zscore': df_zscore['Age'].round(3),
    'Salary': df['Salary'],
    'Salary_zscore': df_zscore['Salary'].round(3),
})

print("\n", comparison_df)
print("\nInterpretation:")
print("- Z-scores near 0: Close to average")
print("- Positive Z-scores: Above average")
print("- Negative Z-scores: Below average")
print("- Now we can fairly compare disparate columns!")
```

## Summary: Key Takeaways

### What We Learned:

1. **Always compare distributions, never raw values**
   - Age and Salary are incomparable on their own scales
   - Distributions show relative patterns meaningfully

2. **Central tendency reveals where data centers**
   - Mean vs Median shows if distribution is skewed
   - Different columns can have very different centers

3. **Variability matters as much as central tendency**
   - Standard deviation shows consistency
   - Coefficient of Variation compares variability fairly

4. **Statistical reasoning guides analysis**
   - Outliers are detected with IQR method
   - Z-scores normalize different scales
   - Correlation shows relationships between variables

5. **EDA is about asking better questions**
   - Why is Diana an outlier in Age?
   - Why does Salary correlate so strongly with Age?
   - Are there any hiring equity concerns?
