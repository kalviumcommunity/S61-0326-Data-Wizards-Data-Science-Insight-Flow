# Applying Vectorized Operations Instead of Python Loops

This notebook shows how to replace simple Python loops with NumPy vectorized operations. The goal is to write code that is clearer, shorter, and better suited to numerical computing.

## 1. What Vectorization Means
Vectorized code applies an operation to an entire array at once instead of iterating item by item in Python. NumPy can execute these operations efficiently because the work happens in optimized array operations rather than explicit loops.

```python
import numpy as np

numbers = np.array([1, 2, 3, 4, 5])

# Loop-based version
doubled_with_loop = []
for number in numbers:
    doubled_with_loop.append(number * 2)

# Vectorized version
doubled_vectorized = numbers * 2

print('Original numbers:   ', numbers)
print('Loop-based result:  ', doubled_with_loop)
print('Vectorized result:  ', doubled_vectorized)
print('Same values?        ', np.array_equal(np.array(doubled_with_loop), doubled_vectorized))
```

## 2. Vectorized Arithmetic
NumPy lets you apply arithmetic directly to arrays. This keeps expressions compact and avoids writing a separate loop for simple element-wise math.

```python
values = np.array([10, 20, 30])
offset = 5

print('values:         ', values)
print('values + 5:     ', values + offset)
print('values * 3:     ', values * 3)
print('values - 2:     ', values - 2)
print('values / 2:     ', values / 2)
```

## 3. Vectorized Comparisons and Conditions
Comparisons on NumPy arrays are also element-wise. The result is a boolean array that you can use for filtering or checking conditions without writing a loop.

```python
temperatures = np.array([18, 22, 27, 31, 16])
hot_days = temperatures > 25

print('temperatures:   ', temperatures)
print('temperatures > 25:', hot_days)
print('Days above 25:  ', temperatures[hot_days])
print('Any above 30?   ', np.any(temperatures > 30))
print('All above 15?   ', np.all(temperatures > 15))
```

## 4. Common Vectorization Mistakes
Vectorization is powerful, but it only works cleanly when the shapes are compatible. If arrays cannot broadcast together, NumPy raises an error. That is a signal to check the dimensions before forcing the operation.

```python
left = np.array([1, 2, 3])
right = np.array([10, 20])

try:
    print(left + right)
except ValueError as error:
    print('Shape mismatch example:')
    print(error)

compatible = np.array([10, 20, 30])
print('Compatible arrays: ', left + compatible)

column = np.array([[1], [2], [3]])
print('Broadcasting example:')
print(column + compatible)
```
