# Inspecting DataFrames with head(), info(), and describe()

This notebook shows how to inspect a pandas DataFrame before cleaning or analysis. The goal is to quickly understand structure, data types, missing values, and basic numeric summaries.

## 1. Load a DataFrame

In a real project, you would load data from a file or database. Here we create a small sample DataFrame so the inspection methods clearly show useful information.

```python
import pandas as pd

df = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106],
    'customer': ['Ava', 'Ben', 'Chloe', 'Dylan', 'Eva', 'Frank'],
    'zone': ['North', 'South', 'North', 'East', None, 'West'],
    'delivery_minutes': [32, 48, 27, 55, 41, 38],
    'delayed': [False, True, False, True, False, False]
})

df
```

## 2. Inspect the First Few Rows with head()

Use `head()` to preview the DataFrame and confirm that the columns and sample values look reasonable. By default, it shows the first five rows.

```python
df.head()
```

What `head()` reveals:
- column names
- sample values
- whether the rows appear aligned correctly
- whether the data looks complete at a glance

## 3. Inspect Structure with info()

Use `info()` to check column types, non-null counts, and memory usage. This is the fastest way to spot missing values and columns that may need cleaning or type conversion.

```python
df.info()
```

What `info()` reveals:
- the number of non-null values in each column
- data types such as `int64`, `object`, and `bool`
- how much memory the DataFrame uses
- where missing data might affect analysis

## 4. Summarize Numeric Columns with describe()

Use `describe()` to get summary statistics for numeric columns. This helps you understand the center, spread, and range of the data before you draw conclusions.

```python
df.describe()
```

What `describe()` reveals:
- `count`: how many numeric values are present
- `mean`: the average
- `std`: the standard deviation
- `min` and `max`: the range
- `25%`, `50%`, and `75%`: useful percentiles for understanding spread

Notice that `describe()` focuses on numeric data by default. It does not replace `info()`, because it does not show missingness or data types in the same way.

## 5. How the Methods Work Together

Use all three methods together before you clean or analyze data:
- `head()` for a visual preview
- `info()` for structure and types
- `describe()` for numeric summaries

This habit helps you catch obvious issues early and makes the next analysis step more reliable.
